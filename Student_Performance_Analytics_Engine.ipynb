```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(7)

print("NumPy Version:", np.__version__)
print("Pandas Version:", pd.__version__)
print("Student Performance Analytics Engine Started!")
```

Generate 500 × 6 Marks Matrix

```python
students = 500
subjects = 6

marks = np.random.normal(65, 12, (students, subjects))
marks = np.clip(marks, 0, 100)
marks = np.round(marks, 2)

print("Marks matrix created successfully!")
print("Matrix shape:", marks.shape)
print("\nFirst 5 students' marks:\n")
print(marks[:5])
```

Calculate Sum, Mean, Median, Standard Deviation and Variance

```python
subject_sum = np.sum(marks, axis=0)
subject_mean = np.mean(marks, axis=0)
subject_median = np.median(marks, axis=0)
subject_std = np.std(marks, axis=0)
subject_var = np.var(marks, axis=0)

print("SUBJECT-WISE STATISTICS")
print("-" * 60)

for i in range(subjects):
    print(f"\nSubject {i + 1}")
    print("Total Marks:", round(subject_sum[i], 2))
    print("Mean:", round(subject_mean[i], 2))
    print("Median:", round(subject_median[i], 2))
    print("Standard Deviation:", round(subject_std[i], 2))
    print("Variance:", round(subject_var[i], 2))
```

Student-wise statistics

```python
student_sum = np.sum(marks, axis=1)
student_mean = np.mean(marks, axis=1)
student_median = np.median(marks, axis=1)
student_std = np.std(marks, axis=1)
student_var = np.var(marks, axis=1)

print("STUDENT-WISE STATISTICS")
print("-" * 50)

for i in range(5):
    print(f"\nStudent {i + 1}")
    print("Total:", round(student_sum[i], 2))
    print("Average:", round(student_mean[i], 2))
    print("Median:", round(student_median[i], 2))
    print("Standard Deviation:", round(student_std[i], 2))
    print("Variance:", round(student_var[i], 2))
```

Calculate Z-Scores and Detect Outliers

```python
subject_mean = np.mean(marks, axis=0)
subject_std = np.std(marks, axis=0)

# Broadcasting: subject statistics apply to every student's marks
z_scores = (marks - subject_mean) / subject_std

print("Z-score matrix shape:", z_scores.shape)
print("\nFirst 5 students' Z-scores:\n")
print(np.round(z_scores[:5], 2))
```

Outlier detection

```python
outlier_mask = np.abs(z_scores) > 2
outlier_rows, outlier_cols = np.where(outlier_mask)
print("OUTLIER DETECTION REPORT")
print("-" * 40)
print("Total unusual subject scores:", len(outlier_rows))
for row, col in zip(outlier_rows[:20], outlier_cols[:20]):
    print(
        f"Student {row + 1}, "
        f"Subject {col + 1}, "
        f"Marks: {marks[row, col]}, "
        f"Z-score: {z_scores[row, col]:.2f}"
    )
```

Rank Students

```python
student_totals = np.sum(marks, axis=1)
student_averages = np.mean(marks, axis=1)

# Descending order: highest total first
ranking = np.argsort(-student_totals)
print("TOP 10 STUDENTS")
print("-" * 50)
for position, student_index in enumerate(ranking[:10], start=1):
    print(
        f"Rank {position}: "
        f"Student {student_index + 1} | "
        f"Total: {student_totals[student_index]:.2f}/600 | "
        f"Average: {student_averages[student_index]:.2f}/100"
    )
```

Add Total Marks Using hstack

```python
totals_column = student_totals.reshape(-1, 1)
averages_column = student_averages.reshape(-1, 1)

# Add total and average columns to original marks matrix
final_data = np.hstack((
    marks,
    totals_column,
    averages_column
))
print("Original matrix shape:", marks.shape)
print("Final matrix shape:", final_data.shape)
print("\nFirst 5 student records:\n")
print(final_data[:5])
```

Create a Pandas DataFrame

```python
columns = [
    "Maths",
    "Physics",
    "Chemistry",
    "Computer_Science",
    "English",
    "Statistics",
    "Total",
    "Average"
]

df = pd.DataFrame(final_data, columns=columns)

df.insert(
    0,
    "Student_ID",
    [f"STU{i:03d}" for i in range(1, students + 1)]
)

df["Rank"] = df["Total"].rank(
    ascending=False,
    method="min"
).astype(int)

df["Outlier_Count"] = outlier_mask.sum(axis=1)

print("Student dataset created!")
print("Dataset shape:", df.shape)

df.head(10)
```

Visualize Student Performance

```python
plt.figure(figsize=(10, 5))

plt.bar(
    columns[:6],
    subject_mean
)

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.ylim(0, 100)
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

Top 10 students ka graph

```python
top10 = df.sort_values(
    "Total",
    ascending=False
).head(10)

plt.figure(figsize=(10, 5))

plt.bar(
    top10["Student_ID"],
    top10["Total"]
)

plt.title("Top 10 Student Performance")
plt.xlabel("Student ID")
plt.ylabel("Total Marks out of 600")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
filename = "student_performance_analytics.csv"

df.to_csv(filename, index=False)

print("CSV file created successfully:", filename)
```
